```python
plt.figure(figsize=(12,5))

sns.countplot(data=df,
              x='fault')

plt.xticks(rotation=90)

plt.show()
```

```python
df.describe()
```

```python
corr=df.corr(numeric_only=True)

plt.figure(figsize=(8,6))

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm')

plt.show()
```
